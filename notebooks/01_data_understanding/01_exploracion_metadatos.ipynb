```python
import requests  # utilizamos esta biblioteca para hcaer solicitudes HTTP desde pythonn

fuentes = {
    "RUES": "c82u-588k",
    "TRM": "mcec-87by",
    "FUENTE_03": "nuke-fusu",
    "SECOP_II": "jbjy-vk9h",
}

for nombre, identificador in fuentes.items():
    url = f"https://www.datos.gov.co/api/views/{identificador}.json"

    respuesta = requests.get(url, timeout=30)
    respuesta.raise_for_status()

    metadatos = respuesta.json()

    print(f"\nFUENTE: {nombre}")
    print(f"Nombre oficial: {metadatos.get('name')}")
    print(f"Columnas: {len(metadatos.get('columns', []))}")

    for columna in metadatos.get("columns", []):
        print(
            columna.get("fieldName"),
            "|",
            columna.get("dataTypeName"),
        )
```

```python
# ============================================================
# 01. CONSULTA EXPLORATORIA DE REGISTROS MERCANTILES - RUES
# ============================================================
# Objetivo:
# Obtener una muestra de 10 registros de la API pública RUES.
#
# Descripción:
# Leer la configuración del proyecto, consultar la API RUES,
# convertir la respuesta JSON en un DataFrame de Pandas y
# visualizar los registros para su exploración inicial.
# ============================================================

from pathlib import Path
import requests
import yaml
import pandas as pd


# 1. Localizar la raíz del proyecto
project_root = Path.cwd()

while not (project_root / "config" / "settings.yaml").exists():
    if project_root == project_root.parent:
        raise FileNotFoundError(
            "No se encontró el archivo config/settings.yaml"
        )
    project_root = project_root.parent


# 2. Cargar la configuración centralizada
with open(
    project_root / "config" / "settings.yaml",
    encoding="utf-8"
) as file:
    config = yaml.safe_load(file)


# 3. Obtener la URL de la fuente RUES
url_rues = config["sources"]["rues"]["url"]


# 4. Consultar una muestra de 10 registros
response = requests.get(
    url_rues,
    params={"$limit": 10},
    timeout=30
)

response.raise_for_status()


# 5. Convertir la respuesta JSON en un DataFrame
df_rues = pd.DataFrame(response.json())


# 6. Visualizar los registros obtenidos
display(df_rues)
```

```python
# ============================================================
# 03. IDENTIFICACIÓN DE COLUMNAS Y TIPOS DE DATOS - RUES
# ============================================================
# Objetivo:
# Identificar todas las columnas de la muestra RUES y los
# tipos de datos que Pandas tiene asociados a cada una.
#
# Descripción:
# Examinar la estructura del DataFrame y generar una tabla
# con el nombre de cada columna y su tipo de dato.
# ============================================================

# 1. Obtener nombres de columnas y tipos de datos
estructura_rues = pd.DataFrame({
    "nombre_columna": df_rues.columns,
    "tipo_dato": df_rues.dtypes.astype(str).values
})

# 2. Agregar numeración para facilitar la lectura
estructura_rues.insert(
    0,
    "numero",
    range(1, len(estructura_rues) + 1)
)

# 3. Mostrar la estructura completa
display(estructura_rues)

# 4. Mostrar el total de columnas identificadas
print(f"Total de columnas identificadas: {len(estructura_rues)}")
```

```python
# ============================================================
# 04. CONSULTA DE VOLUMETRÍA RUES POR AÑO Y MES para RUES
# ============================================================
# Objetivo:
# Consultar cuántos registros existen para un mes específico.
#
# Descripción:
# Permite seleccionar manualmente el año y el mes.
# Consulta únicamente el conteo de registros en Socrata
# y calcula cuántos lotes de descarga se necesitarían.
# ============================================================

import requests
import math

# ------------------------------------------------------------
# 01. PARÁMETROS MODIFICABLES POR EL USUARIO
# ------------------------------------------------------------

ANIO =2026
MES = 1
TAMANO_LOTE = 10000

# ------------------------------------------------------------
# 02. VALIDACIÓN Y CONSTRUCCIÓN DEL FILTRO
# ------------------------------------------------------------

if ANIO not in (2025, 2026):
    raise ValueError("El año debe ser 2025 o 2026.")

if not 1 <= MES <= 12:
    raise ValueError("El mes debe estar entre 1 y 12.")

fecha_inicio = f"{ANIO}{MES:02d}01"

if MES == 12:
    fecha_fin = f"{ANIO + 1}0101"
else:
    fecha_fin = f"{ANIO}{MES + 1:02d}01"

filtro = (
    f"fecha_matricula >= '{fecha_inicio}' "
    f"AND fecha_matricula < '{fecha_fin}'"
)

# ------------------------------------------------------------
# 03. CONSULTA A LA API
# ------------------------------------------------------------

url = "https://www.datos.gov.co/resource/c82u-588k.json"

respuesta = requests.get(
    url,
    params={
        "$select": "count(*) AS total",
        "$where": filtro
    },
    timeout=60
)

respuesta.raise_for_status()

total_registros = int(respuesta.json()[0]["total"])
total_lotes = math.ceil(total_registros / TAMANO_LOTE)

# ------------------------------------------------------------
# 04. RESULTADOS
# ------------------------------------------------------------

print("RESUMEN DE VOLUMETRÍA RUES")
print("-" * 40)
print(f"Período consultado : {ANIO}-{MES:02d}")
print(f"Registros disponibles: {total_registros:,}")
print(f"Tamaño del lote     : {TAMANO_LOTE:,}")
print(f"Lotes necesarios    : {total_lotes}")
```

```python


# ============================================================
#. CONSULTA DE VOLUMETRÍA RUES POR AÑO Y MES para SECOP
# ============================================================
# Objetivo:
# Consultar cuántos registros existen para un mes específico.
#
# Descripción:
# Permite seleccionar manualmente el año y el mes.
# Consulta únicamente el conteo de registros en Socrata
# y calcula cuántos lotes de descarga se necesitarían.
# ============================================================

import requests

ANIO, MES = 2025,2
inicio = f"{ANIO}-{MES:02d}-01"
fin = f"{ANIO + 1}-01-01" if MES == 12 else f"{ANIO}-{MES + 1:02d}-01"
filtro = (
    f"fecha_de_firma >= '{inicio}' AND fecha_de_firma < '{fin}' "
    "AND id_contrato IS NOT NULL"
)

respuesta = requests.get(
    "https://www.datos.gov.co/resource/jbjy-vk9h.json",
    params={"$select": "count(distinct id_contrato) AS contratos", "$where": filtro},
    timeout=180,
)
respuesta.raise_for_status()
contratos = int(respuesta.json()[0]["contratos"])
print(f"Contratos firmados en {ANIO}-{MES:02d}: {contratos:,}")
```
